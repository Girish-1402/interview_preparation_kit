Problem1: Find the largest number

```python
numbers1 = [12,45,23,67,34]

def max_num(numbers:list):
    max_num=0
    for i in numbers:
        if max_num < i:
            max_num = i

    return max_num

res = max_num(numbers1)
print(res)
```

Problem 2: Count even and odd numbers

```python
numbers = [10, 15, 22, 33, 40, 51, 60]

def counter_eve_odd(numbers:list):
    even = 0
    odd = 0
    for num in numbers:
        if num%2==0:
            even+=1
        else:
            odd+=1
    return even,odd

even_cnt,odd_cnt = counter_eve_odd(numbers) 

print(even_cnt)
print(odd_cnt)
```

Problem 3: Find the second-largest unique number

```python
numbers = [10, 5, 20, 8, 20, 15]

def second_large_number(numbers:list):
    numbers = sorted(list(set(numbers)))
    return numbers[-2]

res1 = second_large_number(numbers)
print(res1)
```

<b>Problem4: Count pipeline failures</b>  
Your task: Write Python code that counts how many times each pipeline appears.

```python
pipelines = [
    "SAP_A6P",
    "SAP_F6P",
    "SAP_A6P",
    "SAP_H2P",
    "SAP_F6P",
    "SAP_A6P"
]

def count_of_pipeline(pipelines:list):
    dict={}
    for i in pipelines:
        if i in dict.keys():
            dict[i]=dict[i]+1
        else:
            dict[i]=1

    return dict

res2 = count_of_pipeline(pipelines)
print(res2)
```

<b>Problem 5: Find the first non-repeating number</b>

```python
numbers = [4, 5, 4, 6, 5, 7, 8, 7]

def non_repeating(numbers:list):
    unique_count = 0
    found = False
    for i in range(len(numbers)):
        count = 0
        for j in range(len(numbers)):
            if numbers[i] == numbers[j]:
                count+=1

        if count == 1:
            unique_count +=1
        if unique_count == 1:
            print("first non repeating element",numbers[i])
            found = True
            break

    if not found:
        print("nothing found")


res3 = non_repeating(numbers)

print(res3)
```
