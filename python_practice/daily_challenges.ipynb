Problem1: Find the largest number

```python
numbers1 = [12,45,23,67,34]

def max_num(numbers:list):
    max_num=0
    for i in numbers:
        if max_num < i:
            max_num = i

    return max_num

res = max_num(numbers1)
print(res)
```
